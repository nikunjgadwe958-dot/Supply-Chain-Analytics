```python
# Converting the order date column from text to datetime

df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"], 
    format = "%m/%d/%Y %H:%M"
)
```

```python
# Converting the shipping date column from text to datetime

df["shipping date (DateOrders)"] = pd.to_datetime(
    df["shipping date (DateOrders)"],
    format = "%m/%d/%Y %H:%M"
)
```

```python
# Veryfying Conversion of shippind date 
df["shipping date (DateOrders)"].dtype
```

```python
# Checking if there are Invalid dates created during conversion
df[["order date (DateOrders)",
"shipping date (DateOrders)"]].isnull().sum()
```

```python
# checking values consistency in categorical columns

categorical_columns = [
    "Type",
    "Delivery Status",
    "Order Status",
    "Shipping Mode",
    "Customer Segment",
    "Market"
]

for column in categorical_columns:
    print(f"\n{'='*50}")
    print(column)
    print(df[column].unique())
```

```python
# Checking invalid values in numerical columns
print("Negative Sales:", (df["Sales per customer"]<0).sum())
print("Negative Sales Product:", (df["Product Price"]<0).sum())
print("Negative Actual Shipping Days:", (df["Days for shipping (real)"]<0).sum())
print("Negative Scheduled Shipping Days:", (df["Days for shipment (scheduled)"]<0).sum())
print("Negative Benefit per order:", (df["Benefit per order"]<0).sum())
```

```python
# Extracting the year from the order date 
df["Order Year"] = df["order date (DateOrders)"].dt.year
```

```python
# Checking Year Column
df["Order Year"].value_counts().sort_index()
```

```python
# Extracting the quarter from the order date 
df["Order Quarter"] = "Q" + df["order date (DateOrders)"].dt.quarter.astype(str)
```

```python
# Verifying Quarters Column
df["Order Quarter"].value_counts().sort_index()
```

```python
# Extracting Month Number from order date 
df["Order Month Number"] = df["order date (DateOrders)"].dt.month
```

```python
# Extracting Month Name from Order date
df["Order Month"] = df["order date (DateOrders)"].dt.month_name()
```

```python
# Calculating the profit margin percentage for each order
df["Profit Margin(%)"] = (
    df["Benefit per order"]/df["Sales per customer"]
)*100
```

```python
# Reviewing distribution for profit margin 
df["Profit Margin(%)"].describe()
```

```python
# Calculating deliver Delays in date
df["Delivery Delay"] = (
    df["Days for shipping (real)"] - df["Days for shipment (scheduled)"]
)
```

```python
# Reviewing the distribution of delivery delay
df["Delivery Delay"].describe()
```

```python
# Checking frequency of each delivery date value
df["Delivery Delay"].value_counts().sort_index()
```

```python
# Classifying shipping performance based on delivery delay
df["Shipping Performance"] = np.where(
    df["Delivery Delay"]<0,"Early",
    np.where(df["Delivery Delay"]==0,"On Time","Late")
)
```

```python
# Verifying shipping performance
df["Shipping Performance"].value_counts()
```

```python
# Exploratory Data Analysis(EDA)
# Total Sales by Product Category
category_sales = (
    df.groupby("Category Name")["Sales per customer"].sum().sort_values(ascending=False).round(2)
)

category_sales
```

```python
# Total Profit by product category
category_profit = (
    df.groupby("Category Name", as_index=False)["Benefit per order"].sum().sort_values("Benefit per order", ascending=False).round(2)
)
category_profit
```

```python
# Monthly sales trend
monthly_trend = (df.groupby(
    ["Order Year","Order Month Number","Order Month"],as_index=False
)["Sales per customer"].sum().round(2).sort_values(["Order Year","Order Month Number"])
)
monthly_trend
```

```python
# Top 10 Products by Revenue
top_products = (
    df.groupby("Product Name", as_index=False)["Sales per customer"].sum().sort_values("Sales per customer", ascending=False).head(10).round(2)
)
top_products
```

```python
# Top sales by customer segment
segment_sales = (
    df.groupby("Customer Segment", as_index = False)["Sales per customer"].sum().sort_values("Sales per customer", ascending=False).round(2)
)
segment_sales
```

```python
# Shipping performance by shipping mode
shipping_analysis = (
    df.groupby("Shipping Mode", as_index=False)
      .agg(
          Total_Orders=("Order Id", "count"),
          Avg_Delivery_Delay=("Delivery Delay", "mean"),
          Total_Sales=("Sales per customer", "sum")
      )
      .round(2)
      .sort_values("Total_Orders", ascending=False)
)

shipping_analysis
```
